**1. What is the difference between the two algorithms in terms of the number of steps in the worst case?**

*A:* In a sample of 1 billion elements, in the worst case, linear search will traverse the entire list, resulting in 1 billion comparisons. Binary search, on the other hand, only needs 23 comparisons for the worst case.

*TLDR:* List of 1bn elements | Linear: 1bn comparisons | Binary: 23 comparisons

**2. Why does binary search require the list to be sorted?**

A: Binary search works by dividing the list in half, looking for the middle value, comparing it with the target value, and discarding lower or higher values according to the comparison of the middle value to the target. If the list is not sorted, it is not possible to discard the other values, because if there is no guarantee that the values are in order, there is a possibility that the target value will also be discarded.

**3. In your own words, explain the difference between O(n) and O(log n).**

*A:* Big O notation is used to describe the complexity of an algorithm in execution steps, considering the worst case. O(n) represents the execution of linear search, where the number of steps is proportional to the number of elements. O(log n) represents the execution of binary search, where the number of steps is proportional to the logarithm of the number of elements. This reduces time and increases algorithm efficiency.

*TLDR:* O(n) = linear, steps = n (elements) | O(log n) = binary, steps = log(elements)
