**1. Qual a diferença entre os dois algoritmos em termos de quantidade de passos no pior caso?**

*R:* Em uma amostra de 1 bilhão de elementos, no pior caso a pesquisa linear percorrerá toda a lista, resultando em 1 bilhão de comparações. Já a busca binária necessita somente de 23 comparações para o pior caso.

*TLDR:* Lista de 1b. de elementos | linear: 1b. comparações | binária: 23 comparações

**2. Por que a busca binária exige que a lista esteja ordenada?**

R: A busca binária funciona dividindo a lista em metades buscando o valor do meio, assim comparando com o valor alvo e descartando os valores inferiores ou superiores de acordo com o a comparação do meio para o alvo. Caso a lista não esteja ordenada, não é possível descartar os demais valores, pois se não há garantia de que os valores estão em ordem, há a possibilidade de que o valor alvo será descartado também.

**3. Escreva em suas palavras a diferença entre O(n) e O(log n).**

*R:* A notação Big O é usada para descrever a complexidade de um algoritmo em passos de execução considerando o pior caso. O(n) representa a execução da busca linear, onde o número de passos é proporcional ao número de elementos. Já O(log n) representa a execução da busca binária, onde o número de passos é proporcional ao logaritmo do número de elementos. Reduzindo assim o tempo e aumentando a eficiência do algoritmo.

*TLDR:* O(n) = linear, passos = n (elementos) | O(log n) = binária, passos = log(elementos)
