```python

import numpy
import random
import time
```

```python
#Creating Lists
def generate_list(size):
  best_case = list(range(1, size + 1))
  random_case = random.sample(range(1, size + 1), size)
  worst_case = list(range(size, 0, -1))
  return best_case, random_case, worst_case
```

```python
#Selection Sort Func
def selection_sort(list):
  sample = list.copy()
  print(f"\n--- Selection Sort ---")
  time_start = time.time()
  sorted_list = []
  comparisions = 0 #1*Step counter for notations
  swaps = 0
  for i in range(len(sample)): #2*Append of the lowest value on new sorted list
    lowest = sample[0]
    lowest_idx = 0
    for j in range(1, len(sample)): #3*Search for the lowest value
      if sample[j] < lowest:
        lowest = sample[j]
        lowest_idx = j
      comparisions += 1
    sorted_list.append(sample.pop(lowest_idx)) #2*popping the lowest digit out of original list
    swaps += 1
  time_end = time.time()
  print(f"Sorted: {sorted_list}")
  print(f"Comparisions: {comparisions}")
  print(f"Swaps: {swaps}")
  print(f"Time elapsed: {(time_end - time_start)*1000:.2f}ms")
  return sorted_list
```

```python
#Bubble Sort Func
def bubble_sort(list):
  sample = list.copy()
  print(f"\n--- Bubble Sort ---")
  time_start = time.time()
  comparisions = 0
  swaps = 0
  for i in range(len(sample)): #Registering swap for break in case list is already ordered
    swap = False
    for j in range(0, len(sample) - i - 1): #Starting index, unsorted indexes - sorting for the biggest number
      if sample[j] > sample[j + 1]:
        sample[j], sample[j + 1] = sample[j + 1], sample[j] #Swap for the biggest number until reaching the end
        swap = True
        swaps += 1
      comparisions += 1
    if not swap:
      break
  time_end = time.time()
  print(f"Sorted: {sample}")
  print(f"Comparisions: {comparisions}")
  print(f"Swaps: {swaps}")
  print(f"Time elapsed: {(time_end - time_start)*1000:.2f}ms")
  return

size = 10 #List size modifier for testing ===================================
best_case, random_case, worst_case = generate_list(size)

print(f"\n========= Comparing [Selection Sort] x [Bubble Sort] =========")
print(f"Sorting a list with {size} values")
print(f"\n\n---==== Best Case Scenario ====---")
print(f"Original: {best_case}")
selection_sort(best_case)
bubble_sort(best_case)
print(f"\n\n---==== Random Case Scenario ====---")
print(f"Original: {random_case}")
selection_sort(random_case)
bubble_sort(random_case)
print(f"\n\n---==== Worst Case Scenario ====---")
print(f"Original: {worst_case}")
selection_sort(worst_case)
bubble_sort(worst_case)
```
