## Sorting Algorithms: Selection Sort vs. Bubble Sort

### 1. **Selection Sort**

*   **Concept:** Selection Sort works by repeatedly finding the minimum element from the unsorted part of the list and putting it at the beginning. It maintains two subarrays in a given array: the sorted part and the unsorted part.

*   **How it Works:**
    1.  Start with the first element as the minimum.
    2.  Iterate through the unsorted portion of the array to find the actual minimum element.
    3.  If a smaller element is found, update the minimum.
    4.  After iterating through the entire unsorted portion, swap the found minimum element with the element at the current position in the sorted part.
    5.  Repeat this process for each position until the entire array is sorted.

*   **Time Complexity:**
    *   **Worst-case:** O(n²)
    *   **Best-case:** O(n²)
    *   **Average-case:** O(n²)
    (n is the number of items in the list)

*   **Space Complexity:** O(1) (in-place sorting)

*   **Key Characteristics:**
    *   Simple to understand and implement.
    *   Not efficient for large lists.
    *   Performs minimal swaps (exactly `n-1` swaps), which can be advantageous if writes to memory are costly.
    *   Unstable sorting algorithm (does not preserve the relative order of equal elements).
---
### 2. **Bubble Sort**

*   **Concept:** Bubble Sort repeatedly steps through the list, compares adjacent elements, and swaps them if they are in the wrong order. The pass through the list is repeated until no swaps are needed, which indicates that the list is sorted.

*   **How it Works:**
    1.  Start at the beginning of the list.
    2.  Compare the first two elements. If the first is greater than the second, swap them.
    3.  Move to the next pair of elements and repeat the comparison and swap.
    4.  Continue this process until the end of the list. After the first pass, the largest element will have "bubbled" to the end.
    5.  Repeat the entire process for the remaining unsorted portion of the list, reducing the range of comparisons by one element each time, until no more swaps are necessary.

*   **Time Complexity:**
    *   **Worst-case:** O(n²)
    *   **Best-case:** O(n) (if the list is already sorted)
    *   **Average-case:** O(n²)

*   **Space Complexity:** O(1) (in-place sorting)

*   **Key Characteristics:**
    *   Very simple to understand and implement.
    *   Inefficient for large lists.
    *   Can be optimized to detect if the list is already sorted, stopping early (best-case O(n)).
    *   Stable sorting algorithm (preserves the relative order of equal elements).
