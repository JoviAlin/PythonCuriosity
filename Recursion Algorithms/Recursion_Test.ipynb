```python
import time
import tracemalloc
```

```python
#Defining the fibonacci pure recursion function
def fibonacci_pure(n):
  if n <= 1: #Base problem
    return n
  else:
    return fibonacci_pure(n-1) + fibonacci_pure(n-2) #The fibonacci formula/Recursive problem
```

```python
#Defining the fibonacci cache recursion function
def fibonacci_cache(n, cache={}):
  if n <= 1: #Base problem
    return n
  if n in cache:
    return cache[n]

  cache[n] = fibonacci_cache(n-1) + fibonacci_cache(n-2)
  return cache[n]
```

```python
#Defining the test result function
def run_test(n):
  #Results and performance for Pure Fibonnaci
  timestart=time.time()
  tracemalloc.start()
  print(f"\n===== Pure Fibonacci Sequence To The {n}° Term ====\n")
  for i in range(n+1):
    print(f"F({i}) = {fibonacci_pure(i)}")

  #Performance Variables
  timeend=time.time()
  memnow, mempeak = tracemalloc.get_traced_memory()
  tracemalloc.stop()

  print(f"\nTime elapsed: {(timeend-timestart)*1000:.2f}ms")
  print(f"Memory Use Peak: {mempeak / 1024:.2f}KB")
  print(f"Memory Use Now: {memnow / 1024:.2f}KB")

  #Results and performance for Cache  Fibonnaci
  timestart=time.time()
  tracemalloc.start()
  print(f"\n===== Cache Fibonacci Sequence To The {n}° Term ====\n")
  for i in range(n+1):
    print(f"F({i}) = {fibonacci_cache(i)}")

  #Performance Variables
  timeend=time.time()
  memnow, mempeak = tracemalloc.get_traced_memory()
  tracemalloc.stop()

  print(f"\nTime elapsed: {(timeend-timestart)*1000:.2f}ms")
  print(f"Memory Use Peak: {mempeak / 1024:.2f}KB")
  print(f"Memory Use Now: {memnow / 1024:.2f}KB")

#Test variable
n=int(input("Base Variable for Testing: "))
run_test(n)
```
