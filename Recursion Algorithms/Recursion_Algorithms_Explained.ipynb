#**Recursion Algorithms**

A **recursive algorithm** is a method for solving a problem where the solution depends on *solving smaller instances of the same problem*. In other words, a function calls itself, either directly or indirectly, until it reaches a condition simple enough to solve without further recursion.


###**Every recursive algorithm relies on two essential components:**

- **The base case** is the simplest version of the problem — one that can be solved directly, without any further recursive calls. It's what stops the recursion from running forever; without it, the function would keep calling itself endlessly, eventually causing a stack overflow.

- **The recursive case** is the part where *the function calls itself again*, but with a smaller or simpler version of the original problem. Each call must move the problem closer to the base case, gradually reducing its complexity until it's simple enough to be solved directly.

###**Pure Recursion vs. Recursion with Memoization (Caching)**

**Pure recursion** solves a problem by breaking it into subproblems and calling itself repeatedly, without storing any of the results it computes along the way. *Every time the same subproblem appears again, it gets recalculated from scratch.*

**Recursion with memoization** works the same way, but it stores ("caches") the result of each subproblem the first time it's solved. *If that same subproblem comes up again, the function simply looks up the stored answer instead of recomputing it.*

---
####1. **Why isn't pure recursion ideal?**

>The main problem with pure recursion is that it often recalculates the same subproblems many times over, especially in problems where subproblems overlap.

A classic example is the Fibonacci sequence. To compute the 5th Fibonacci number, the function needs to compute the 4th and 3rd numbers. But computing the 4th number also requires computing the 3rd number again — *so that subproblem ends up being solved twice.* This duplication grows exponentially as the input size increases: smaller subproblems get recalculated over and over, sometimes hundreds or thousands of times for relatively small inputs.

##### **This leads to:**

- *Exponential time complexity* — the number of calculations roughly doubles with each additional step, making the algorithm extremely slow for larger inputs.
Wasted computation — the same work is repeated unnecessarily instead of being reused.
- *Poor scalability* — the function becomes impractical for anything beyond small input sizes.

---
####2. **What is the time complexity of each approach?**

> - Pure recursion: O(2ⁿ) — exponential
> - Recursion with memoization: O(n) — linear

#####**Why is recursion exponential?**
**Each call to fib(n) triggers two more calls (fib(n-1) and fib(n-2))**, which each trigger two more, and so on. This forms a call tree that roughly *doubles in size at each level*, and the depth of that tree is n. Since the same subproblems (like fib(3)) get recalculated repeatedly across different branches of the tree, the total number of calls grows exponentially with n.

#####**Why is memoization linear?**

Since **each subproblem (fib(0) through fib(n)) is calculated only once** and *then stored in a cache*, any later request for that value is simply looked up — no recalculation needed. Since there are only n distinct subproblems, *and each is solved just once*, the total time grows proportionally to n.

---
####3.**Is it possible to determine which is more efficient using other parameters (time or memory, for example)?**

Lets take a look at space (memory) complexity, and the trade-off between the two tells the full story.

#####**Time comparison:**

- Pure recursion	O(2ⁿ)

- Memoized recursion	O(n)

>In terms of raw speed, memoization is drastically more efficient.

*For example, computing fib(30):*

Pure recursion makes roughly 2,692,537 calls.

Memoized recursion makes only 31 calls (one per unique subproblem).

>That gap grows explosively as n increases, *so memoization wins decisively on time.*


#####**Space comparison**
- Pure recursion	O(n) — just the call stack depth
- Memoized recursion	O(n) — call stack plus the cache storing n results

**Here's the nuance:** pure recursion only uses space for the recursion call stack (depth n), since it doesn't store any results. Memoization uses that same call stack space, plus extra space for the cache (a dictionary or array holding all n computed values).

>In terms of memory alone, pure recursion has a slightly smaller footprint

both are still O(n), so the difference is minor, not a different complexity class.


---
#####**Which is better?**
If the priority is speed, memoization is overwhelmingly better — the time savings (exponential → linear) vastly outweigh the modest extra memory cost.

If the priority is minimizing memory usage in a system with severe memory constraints (and speed isn't critical), pure recursion *could* be preferred, since it avoids the cache — but this is rarely worth the *massive time penalty* in practice.
